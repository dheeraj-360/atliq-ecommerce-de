```python
%sql

CREATE TABLE IF NOT EXISTS atliq_dl.silver.dq_log (
    run_date STRING,
    table_name STRING,
    rule STRING,
    rows_affected BIGINT
)
USING DELTA;
```

```python
%sql
SELECT
    table_name,
    rule,
    rows_affected,
    run_date
FROM atliq_dl.silver.dq_log
ORDER BY table_name, rule;
```
